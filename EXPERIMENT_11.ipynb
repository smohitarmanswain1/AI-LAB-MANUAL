```python
def solve_queens(n):
    board = [-1] * n  # Initialize board with -1 (no queen placed)
    result = []      # List to store all valid solutions

    def is_valid(row, col):
        # Check if placing a queen at (row, col) is valid
        for i in range(row):
            # Check for queens in the same column or diagonals
            if board[i] == col or \
               abs(row - i) == abs(col - board[i]):
                return False
        return True # If no conflicts, it's valid

    def backtrack(row):
        # Base case: if all queens are placed
        if row == n:
            result.append(list(board)) # Add a copy of the current board to results
            return

        # Try placing a queen in each column of the current row
        for col in range(n):
            if is_valid(row, col):
                board[row] = col       # Place queen
                backtrack(row + 1)     # Recurse for the next row
                board[row] = -1        # Backtrack: remove queen (optional, as it will be overwritten)

    backtrack(0) # Start the backtracking from the first row (row 0)
    return result

# Solve for N=8 queens
solutions = solve_queens(8)

# Print all solutions
for solution in solutions:
    print(f"Solution {solutions.index(solution) + 1}:")
    for row in range(8):
        line = ""
        for col in range(8):
            if solution[row] == col:
                line += "Q "
            else:
                line += "- "
        print(line)
    print("\n") # Add an extra newline between solutions
```
