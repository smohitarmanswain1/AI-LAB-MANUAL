```python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def minimax(node, depth, isMax):
    if depth == 0 or (node.left is None and node.right is None):
        return node.value

    if isMax:
        # Ensure to handle cases where a child might be None
        left_score = minimax(node.left, depth - 1, False) if node.left else float('-inf')
        right_score = minimax(node.right, depth - 1, False) if node.right else float('-inf')
        return max(left_score, right_score)
    else:
        # Ensure to handle cases where a child might be None
        left_score = minimax(node.left, depth - 1, True) if node.left else float('inf')
        right_score = minimax(node.right, depth - 1, True) if node.right else float('inf')
        return min(left_score, right_score)

# Create game tree
root = TreeNode(0)

root.left = TreeNode(0)
root.right = TreeNode(0)

root.left.left = TreeNode(3)
root.left.right = TreeNode(5)

root.right.left = TreeNode(2)
root.right.right = TreeNode(9)

# Run Minimax
result = minimax(root, 2, True)

print("Optimal value:", result)
```
