```python
# 8-Puzzle Problem

# Initial state
puzzle = [
    [1, 2, 3],
    [5, 6, 0],
    [7, 8, 4]
]

# Print the initial state
for row in puzzle:
    print(*row)

print()

# Move 0 to the left
puzzle[1][2], puzzle[1][1] = puzzle[1][1], puzzle[1][2]

for row in puzzle:
    print(*row)

print()

# Move 0 down
puzzle[1][1], puzzle[2][1] = puzzle[2][1], puzzle[1][1]

for row in puzzle:
    print(*row)

print()

# Move 0 to the left
puzzle[2][1], puzzle[2][0] = puzzle[2][0], puzzle[2][1]

for row in puzzle:
    print(*row)
```
